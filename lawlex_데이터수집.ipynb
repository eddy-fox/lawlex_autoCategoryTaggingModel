```python
import time
import random
import requests
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import pandas as pd
import subprocess
from selenium.webdriver.support.ui import WebDriverWait 
from selenium.webdriver.support import expected_conditions as EC
import json
import os
import urllib.parse
```

```python
options = Options()
options.add_argument("--start-maximized")  
options.add_experimental_option("detach", True)

# 크롬 드라이버 설정
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)



file = 'lawtalk_data.csv'

def get_data(page_number):

    numbers = []
    interests = []
    titles = []

    url = f'https://www.lawtalk.co.kr/qna/{page_number}'
    headers = {'User-Agent': 'Mozila/5.0'}
    driver.get(url)
    time.sleep(4)  

    soup = BeautifulSoup(driver.page_source, 'html.parser')


    elements = soup.find_all('div', class_='question directive qna-detail-page expanded')
    
    if not elements :
        return

    for element in elements:
        interest = element.find_all('span')[0].text.strip()
        title = element.find_all('h1')[0].text.strip()
        numbers.append(page_number)
        interests.append(interest)
        titles.append(title)
        
    if interests :
        df_category = pd.DataFrame({
            'number' : numbers,
            'interest': interests,
            'title': titles
        })

        is_file_exists = os.path.exists(file)

        df_category.to_csv(
            file,
            index=False,
            mode='a',
            header=not is_file_exists,
            encoding='utf-8-sig'
        )
        print(f'{page_number} 저장완료')


for page in range(443349, 560000):
    get_data(page)


```
