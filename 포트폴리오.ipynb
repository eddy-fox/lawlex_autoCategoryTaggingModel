# A. 프로젝트 개요

## 프로젝트명
lawlex project의 상담게시판 게시글 자동 카테고리 태깅 모델

## 프로젝트를 들어가며

법률자문서비스 lawlex project를 개발하며 상담게시판에서 카테고리별로 질문글을 분류할 필요가 있었다. 하지만 법률의 종류가 많고, 법에 대해 잘 알지 못하는 일반 유저들이 실제 법률의 종류를 구분하여 카테고리를 선택하는 데 어려움이 있을 것으로 판단했다. 또한, 세부 카테고리가 많아 직접 카테고리를 설정하는 데 있어 시각적 어려움이 있을 것으로 예상했다. 때문에 벤치마킹했던 lawtalk의 시스템에서 차용하여 유저가 상담글의 제목을 입력할 시 제목을 분석하여 자동으로 카테고리를 분류, 추천하는 모델을 개발했다. 

### 사용된 개발환경, 패키지, 툴, 응용프로그램

프로그래밍 언어 : Python3(3.11)<br>
데이터 수집 : Selenium, BeautifulSoup, request<br>
데이터 전처리 : pandas, numpy<br>
머신러닝 패키지 : torch, transformer, skikit-learn<br>
서버구축 패키지 : fastapi<br>
개발환경 : vscode, google colab<br>

# 프로젝트에 고려할 수 있는 AI 서비스 & API (유, 무료)

- Kobert-tokenizer : 한국어 텍스트(뉴스/게시글/유저 입력)를 KoBERT 모델 학습·추론용 토큰/ID로 변환하는 전처리 도구(라이브러리)
- lawtalk(website) : 게시글 제목 및 카테고리 정보
- google colab : GPU 활용 학습 환경

## 프로젝트의 목표
게시글의 제목을 기반으로 세부카테고리를 추천하는 시스템을 개발한다. 유저가 추천받은 세부카테고리 중 하나를 선택하면 세부카테고리가 맵핑된 주요카테고리에 해당 글이 저장된다. 이를 통해 유저의 질문에 보다 정확한 답변을 제공할 수 있으며 비슷한 카테고리의 글을 모아 확인할 수 있는 시각적 편의성을 제공한다

[딥러닝 Code File (Google Colab)](https://drive.google.com/file/d/1REh-17JQqqCcZ9JPgCTMccTjULrRc-Ll/view?usp=sharing)

### 4.2 결론

처음 학습을 시작할 때 까지만 해도 카테고리와 글제목이 정해져 있어 정확도가 최소 80%는 당연히 나올 줄 알았다. 하지만 일주일간의 학습 결과, train data와 test data 모두 정확도가 4%로 머물렀다. 다시 처음부터 코드를 살펴보니 학습 스케줄에서 cosine을 사용했고, 누적그라드가 반영되지 않았기 문에 파라미터 업데이트가 비효율적이었다는 것을 확인했다. 그래서 스케줄러를 linear warmup/decay로 변경한 후 직관적으로 학습 스케줄과 웜업 비율을 설정하니 초기 정확도가 27%까지 향상되었다. 이후 희소한 클래스가 더 자주 뽑히도록 가중치를 설정하면서 샘플링을 조정하지만 loss의 class_weight는 제거하는 방법을 통해 train data 정확도 60%와 test data 정확도 50%로 수치를 끌어올렸다. 50%의 정확도로 원하는 결과가 나올 수 있을지 걱정했지만 테스트해보니 만족스러운 결과가 나왔다. 

# F.마무리

머신러닝 모델만 개발해보다가 이번에 처음으로 딥러닝 모델을 직접 구축하면서 예상보다 시간이 오래 걸렸지만, 그만큼 딥러닝의 전체 학습 흐름과 학습률 스케줄러 등 주요 설정, 그리고 관련 API 활용까지 폭넓게 익힐 수 있었다. 여러가지가 아쉬웠지만 그 중에서도 초기 성능 저하를 하이퍼파라미터 튜닝만으로 해결할 수 있다고 판단해 불필요하게 많은 시간과 GPU 자원을 소모한 점이 가장 아쉽다. 이번 경험을 통해 정확도가 비정상적으로 낮거나 학습이 진행되지 않을 때는 튜닝에 앞서 학습률, 스케줄러, step 계산 등 기본 설정부터 점검해야 한다는 교훈을 얻었다.
